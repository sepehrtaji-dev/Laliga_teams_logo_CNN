```python
import os
import shutil
from sklearn.model_selection import train_test_split
input_folder = r"Laliga teams"
output_folder = r"split_data"

classes = os.listdir(input_folder)
splits = ["train", "val", "test"]

for split in splits:
    for cls in classes:
        os.makedirs(os.path.join(output_folder, split, cls), exist_ok=True)

for cls in classes:
    cls_path = os.path.join(input_folder, cls)
    images = [f for f in os.listdir(cls_path) if f.lower().endswith((".png", ".jpg", ".jpeg"))]

    train_imgs, temp_imgs = train_test_split(images, test_size=0.2, random_state=42)
    val_imgs, test_imgs = train_test_split(temp_imgs, test_size=0.5, random_state=42)

    for img_list, split in [(train_imgs, "train"), (val_imgs, "val"), (test_imgs, "test")]:
        for img in img_list:
            src = os.path.join(cls_path, img)
            dst = os.path.join(output_folder, split, cls, img)
            shutil.copy(src, dst)

    print(f"{cls}: {len(train_imgs)} train, {len(val_imgs)} val, {len(test_imgs)} test")

print("Done splitting!")
```
